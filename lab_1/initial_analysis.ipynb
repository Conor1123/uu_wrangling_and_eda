```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

pd.set_option('display.float_format', lambda x: f'{x:,.0f}')
df = pd.read_csv('ForeignGifts_edu.csv',low_memory=False)
df.head()
```

```python
#Question 1.2
'''The three gift types are contracts, monetary gifts, and real estate. Contract and monetary gifts are the most common, with contract being the
most, and the average gift amounts for both are around the same. Real estate is much less common, only 11 gifts, but has a much higher average
value, though with so few gifts that number is not very reliable.'''

display(df['Gift Type'].unique())
display(df['Gift Type'].value_counts())
df.groupby('Gift Type')['Foreign Gift Amount'].describe()
```

```python
#Question 1.3.1
#Qatar gives the most money

df.groupby('Country of Giftor')['Foreign Gift Amount'].sum().sort_values(ascending=False)
```

```python
#Question 1.3.2
#England recieves the most number of gifts

df.groupby('Country of Giftor')['Foreign Gift Amount'].count().sort_values(ascending=False)
```

```python
#Question 1.3.3
#Bermuda has the highest mean gift amount while Antigua has the highest meadian.

display(df.groupby('Country of Giftor')['Foreign Gift Amount'].mean().sort_values(ascending=False))
df.groupby('Country of Giftor')['Foreign Gift Amount'].median().sort_values(ascending=False)
```

```python
#Question 1.3.4
#Carnegie Mellon University recieved the most amount of money.

df.groupby('Institution Name')['Foreign Gift Amount'].sum().sort_values(ascending=False)
```

```python
#Question 1.3.5
#University of California recieved the most number of gifts.

df.groupby('Institution Name')['Foreign Gift Amount'].count().sort_values(ascending=False)
```

```python
#Question 1.3.6
#The mean gift amount is ~$590K while the median is ~$95K. and difference is due to a few very large gifts skewing the mean.

display(round(df['Foreign Gift Amount'].mean()))
round(df['Foreign Gift Amount'].median())
```

```python
#Question 1.3.7
#The largest flow from county to institution is from Qatar to Cornell.

df.groupby(['Country of Giftor', 'Institution Name'])['Foreign Gift Amount'].sum().sort_values(ascending=False)
```

```python
#Question 1.3.8
#The largest indivdual gifter is the Qatar Foundation.

df.groupby('Giftor Name')['Foreign Gift Amount'].sum().sort_values(ascending=False)
```
