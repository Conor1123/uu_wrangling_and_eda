```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

pd.set_option('display.float_format', lambda x: f'{x:,.0f}')
df = pd.read_csv('ForeignGifts_edu.csv',low_memory=False)
df.head()
```

```python
df[df['Institution Name'].str.contains('Virginia', case=False)]['Institution Name'].unique()
```

```python
uva = df[df['Institution Name'] == 'University of Virginia']
display(uva['Gift Type'].value_counts())
display(uva.groupby('Country of Giftor')['Foreign Gift Amount'].sum().sort_values(ascending=False).head(15))
display(uva.groupby('Country of Giftor')['Foreign Gift Amount'].count().sort_values(ascending=False).head(15))
display(uva.groupby('Giftor Name')['Foreign Gift Amount'].sum().sort_values(ascending=False).head(15))
display(uva.groupby('Giftor Name')['Foreign Gift Amount'].count().sort_values(ascending=False).head(15))
```

```python
peers = ['University of Virginia', 'University of North Carolina - Chapel Hill',
         'Duke University', 'University of Michigan - Ann Arbor', 'Emory University',
         'University of Pittsburgh']
df[df['Institution Name'].isin(peers)].groupby('Institution Name')['Foreign Gift Amount'].agg(['count', 'sum', 'mean']).sort_values('sum', ascending=False)
```

```python
sub = df[df['Institution Name'].isin(peers)]
tc = sub.groupby(['Institution Name', 'Country of Giftor'])['Foreign Gift Amount'].sum().sort_values(ascending=False)
display(tc.head(15))
```

Based on the analysis, UVA should focus on three strategies to grow foreign giving. First, UVA should restart giving from its existing global health partners in Bangladesh and Tanzania. These partners gave gifts multiple times in the past, so bringing these gifts back is the easiest place to start. Second, UVA should try to get more money from AstraZeneca, a Swedish drug company that already funds UVA. UVA has received about 3.2 million dollars from them across two contracts, while Duke has received about 42.5 million dollars across six contracts with the same company. This shows AstraZeneca is willing to give much more to a school like UVA. Third, UVA should look for money in countries that its peers get money from but UVA does not. Duke received 137 million dollars from France, and Michigan received large gifts from Japan, China, and Korea, and none of these countries show up in UVA's list of giving countries at all. Overall, the data shows that most large foreign gifts come from organizations and companies instead of individual people, so UVA should build research and business relationships overseas rather than looking for individual international donors.
